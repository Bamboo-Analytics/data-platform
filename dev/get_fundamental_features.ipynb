there's eps surprise quarterly historical data since 2020 (eps surprise per each quarter earnings), but not the same for the other financials like balance sheet, income, cashflow

EPS surprise is available quarterly because eps does not come from the balance sheet / cashflow statements, but from the earnings calendar endpoints on Yahoo, storing earnings estimates & reported eps per quarter.

in short, earnings dates can be collected historically further back than the balance sheet or cashflow statements. EPS can be  derived directly from these earnings dates.
